```python
import pandas as pd
import yfinance as yf

russell_ticker = pd.read_excel('rusellticker.xlsx')
russell_ticker_list = [str(x) for x in russell_ticker['ticker'] if x == x]
data = yf.download(russell_ticker_list, start='2023-01-01')
data.index = pd.to_datetime(data.index)
data = data.swaplevel(0, 1, axis=1).sort_index(axis=1)
data
```

```python
col =data.columns[data.isna().any()].tolist()
col_names = []
for c in col:
    if c[0] not in col_names:
        col_names.append(str(c[0]))
    else:
        continue
col_names
```
