```python
import wrds
import pandas as pd
db = wrds.Connection()

tickers_df = pd.read_excel("RIY Index Constituents.xlsx")
unique_tickers = set()
cols = tickers_df.columns
for col in cols:
    tickers_for_year = tickers_df[col].unique()
    for ticker in tickers_for_year:
        unique_tickers.add(ticker.split()[0].strip().upper())
print(f"Unique tickers: {len(unique_tickers)}")

years = [x for x in range(2000, 2024)]

count = 0
empty_tickers = []
with pd.HDFStore("all_options_data.h5") as store:
    for ticker in unique_tickers:
        ticker_data = []
        for year in years:
            data = db.raw_sql(
                f"""
                select 
                    b.ticker,
                    a.date, 
                    a.exdate,
                    a.cp_flag,
                    a.strike_price,
                    a.best_bid,
                    a.best_offer,
                    a.open_interest,
                    a.impl_volatility,
                    a.delta,
                    a.gamma,
                    a.theta,
                    a.vega,
                    a.volume
                from optionm_all.opprcd{year} a, optionm.securd b
                where a.secid = b.secid
                and b.ticker = '{ticker}'
                and a.exdate - a.date <= 14
                """, date_cols=["date", "exdate"])
            if not data.empty:
                ticker_data.append(data)
        
        count += 1
        if ticker_data:
            full_data = pd.concat(ticker_data, ignore_index=True)
            full_data.sort_values(by=["date"], inplace=True)
            full_data["ticker"] = full_data["ticker"].astype("category")
            store.put(ticker, full_data, format="table")
            print(f"Stored {ticker}, count: {count}")
        else:
            print(f"No data: {ticker}, count: {count}")
            empty_tickers.append(ticker)

print("Completed")
```
